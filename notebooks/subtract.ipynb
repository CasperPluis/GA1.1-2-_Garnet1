```python
def subtract(a, b):
    result = a - b
    return result

subtract(10, 4)
```

```python
def distance_from_zero(x):
    if x >= 0:
        return subtract(x, 0)
    else:
        return subtract(0, x)

print(distance_from_zero(6))    # 6
print(distance_from_zero(-4))   # 4
print(distance_from_zero(0))    # 0
```
